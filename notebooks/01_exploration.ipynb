Path permet de manipuler des chemins de fichiers.
Et glob() recherche les fichiers correspondant à un motif.

```python
from pathlib import Path
dossier = Path("../data/raw")

fichiers = list(dossier.glob("StormEvents_details*d2024*.csv.gz"))

print(fichiers)
```

```python
df = pd.read_csv(fichiers[0], compression="gzip")

df.head()
```

```python
# Dimensions du jeu de données
print("Dimensions :", df.shape)

# Liste des colonnes
print("\nColonnes :")
print(df.columns.tolist())

# Types des variables
print("\nInformations générales :")
df.info()
```
